## 3. Data types

**Why this matters:** `order_date` and `order_time` are read in as plain text
(`object`/`str`), not real dates — pandas has no way to know they're dates
unless we tell it. `pizza_id`, `order_id`, and `quantity` are read as
`float64` even though every value is a whole number, because pandas' CSV
reader defaults numeric columns to float. Neither of these is wrong, exactly
— but neither is *usable* yet either. That's the job of Notebook 02.

```python
df.dtypes
```

```python
df.info()
```

## 4. Descriptive statistics

```python
df.describe()
```

`describe()` only summarizes numeric columns by default. For a quick look at the categorical (text) columns, use `describe(include='object')`:

```python
df.describe(include='object')
```

## 5. Interrogating categorical columns: `value_counts`, `unique`, `nunique`

These three methods are how you get to know a column you've never seen before.

```python
df['pizza_category'].value_counts()
```

```python
df['pizza_size'].unique()
```

```python
print("Distinct pizza flavors (pizza_name):", df['pizza_name'].nunique())
print("Distinct SKUs (pizza_name_id):", df['pizza_name_id'].nunique())
print("Distinct orders:", df['order_id'].nunique())
```
